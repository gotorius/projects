```python
from pathlib import Path
from PIL import Image

base_dir = Path("/mnt/data1/Public/MedImages/DermMel/train_sep")

classes = ["Melanoma", "NotMelanoma"]
for cls in classes:
    files = list((base_dir / cls).glob("*.*"))
    print(f"{cls}: {len(files)} images")

# 1枚だけ開いて情報を確認
sample_path = list((base_dir / "Melanoma").glob("*.jpeg"))[0]
sample = Image.open(sample_path)
print("Image path:", sample_path)
print("Image size:", sample.size)
print("Mode:", sample.mode)
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

base_dir = Path("/mnt/data1/Public/MedImages/DermMel/train_sep")

classes = ["Melanoma", "NotMelanoma"]

for cls in classes:
    files = list((base_dir / cls).glob("*.*"))  # 拡張子問わず
    print(f"{cls}: {len(files)} images")

    # 上位5枚を表示
    sample_files = files[:5]
    fig, axes = plt.subplots(1, len(sample_files), figsize=(15, 3))
    fig.suptitle(f"{cls} samples", fontsize=14)

    for ax, path in zip(axes, sample_files):
        img = Image.open(path)
        ax.imshow(img)
        ax.set_title(path.name[:12], fontsize=8)
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```
