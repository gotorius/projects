# 必要なライブラリのインポート

画像取得・表示に必要なライブラリ（torchvision, PIL, matplotlibなど）をインポートします。

```python
# 必要なライブラリのインポート
import os
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import datasets, transforms
```

# ImageNetデータセットから画像を取得

torchvision.datasets.ImageNetを使い、ImageNetから任意の画像を一枚ロードします。

```python
# ImageNetデータセットから画像を取得
# ImageNetのデータパスを指定してください
imagenet_root = '/mnt/data1/gotou/kaggle/imagenet-256'  # 例: ImageNetのrootディレクトリ

# 前処理（必要に応じて変更）
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

# データセットの作成
dataset = datasets.ImageFolder(root=imagenet_root, transform=transform)

# 画像とラベルを一枚取得
img_tensor, label = dataset[0]

# 元画像も取得（表示用）
original_img_path = dataset.imgs[0][0]
original_img = Image.open(original_img_path)
```

# 画像の表示

取得した画像をmatplotlibで表示します。

```python
# 画像の表示
plt.figure(figsize=(4, 4))
plt.imshow(original_img)
plt.title(f'Label: {label}')
plt.axis('off')
plt.show()
```
